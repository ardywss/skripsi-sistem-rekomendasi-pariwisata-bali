# **Import Library**

```python
import pandas as pd
import numpy as np

from google.colab import files

pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', None)
pd.set_option('display.width', None)
```

# **Upload Dataset dari Desktop**

```python
uploaded = files.upload()
```

# **Load Dataset**

```python
file_name = list(uploaded.keys())[0]

df = pd.read_excel(file_name)

print("Dataset berhasil dimuat.")
```

# **Informasi Dataset**

```python
print("="*60)
print("INFORMASI DATASET")
print("="*60)

print(f"Jumlah Data   : {df.shape[0]:,}")
print(f"Jumlah Atribut: {df.shape[1]}")

print("\nNama Atribut:")
display(pd.DataFrame(df.columns, columns=["Atribut"]))

print("\nTipe Data:")
display(df.dtypes.reset_index().rename(columns={
    "index":"Atribut",
    0:"Tipe Data"
}))
```

```python
print("5 Data Pertama")

display(df.head())
```

# **Identifikasi Missing Value**

```python
missing = pd.DataFrame({
    "Atribut": df.columns,
    "Jumlah Missing Value": df.isnull().sum().values,
    "Persentase (%)": (
        df.isnull().sum().values /
        len(df) * 100
    ).round(2)
})

print("="*60)
print("MISSING VALUE")
print("="*60)

display(missing)
```

# **Identifikasi Duplikasi Baris**

```python
duplicate_rows = df.duplicated().sum()

print("="*60)
print("DUPLIKASI BARIS")
print("="*60)

print(f"Jumlah duplikasi baris: {duplicate_rows}")
```

# **Identifikasi Place Name dengan Banyak Place ID**

```python
multiple_place_id = (
    df.groupby("place_name")["place_id"]
      .nunique()
      .reset_index(name="Jumlah Place ID")
)

multiple_place_id = multiple_place_id[
    multiple_place_id["Jumlah Place ID"] > 1
]

print("="*60)
print("PLACE NAME DENGAN LEBIH DARI SATU PLACE ID")
print("="*60)

print(f"Jumlah Place Name: {len(multiple_place_id)}")

display(multiple_place_id.head(20))
```

# **Identifikasi User-Item Ganda**

```python
duplicate_user_item = (
    df.groupby(["user_id","place_id"])
      .size()
      .reset_index(name="Jumlah")
)

duplicate_user_item = duplicate_user_item[
    duplicate_user_item["Jumlah"] > 1
]

print("="*60)
print("USER-ITEM DENGAN LEBIH DARI SATU RATING")
print("="*60)

print(f"Jumlah pasangan user-item: {len(duplicate_user_item)}")

display(duplicate_user_item.head(20))
```

# **Ringkasan Kondisi Awal Dataset**

```python
print("="*70)
print("RINGKASAN KONDISI AWAL DATASET")
print("="*70)

print(f"Jumlah Data                     : {len(df):,}")
print(f"Jumlah Missing Value            : {df.isnull().sum().sum():,}")
print(f"Jumlah Duplikasi Baris          : {duplicate_rows:,}")
print(f"Place Name > 1 Place ID         : {len(multiple_place_id):,}")
print(f"User-Item > 1 Rating            : {len(duplicate_user_item):,}")
```

# **Seleksi Atribut**

```python
required_columns = [
    "user_id",
    "place_id",
    "rating",
    "place_name",
    "review_date",
    "source"
]

df_clean = df[required_columns].copy()

print("="*60)
print("SELEKSI ATRIBUT")
print("="*60)

print(f"Jumlah atribut sebelum seleksi : {df.shape[1]}")
print(f"Jumlah atribut sesudah seleksi : {df_clean.shape[1]}")

display(df_clean.head())
```

# **Missing Value pada 6 atribut utama**

```python
missing_required = pd.DataFrame({
    "Atribut": df_clean.columns,
    "Jumlah Missing Value": df_clean.isnull().sum().values,
    "Persentase (%)": (
        df_clean.isnull().sum().values /
        len(df_clean) * 100
    ).round(2)
})

print("="*60)
print("MISSING VALUE PADA ATRIBUT UTAMA")
print("="*60)

display(missing_required)
```

# **Menghapus Missing Value**

```python
jumlah_sebelum = len(df_clean)

df_clean = df_clean.dropna(
    subset=required_columns
).reset_index(drop=True)

jumlah_sesudah = len(df_clean)

print("="*60)
print("PENGHAPUSAN MISSING VALUE")
print("="*60)

print(f"Jumlah data sebelum : {jumlah_sebelum:,}")
print(f"Jumlah data sesudah : {jumlah_sesudah:,}")
print(f"Jumlah data dihapus : {jumlah_sebelum-jumlah_sesudah:,}")
```

# **Normalisasi place_id**

```python
df_clean["review_date"] = pd.to_datetime(df_clean["review_date"])

canonical_place = (
    df_clean
    .sort_values("review_date")
    .groupby("place_name")
    .tail(1)[["place_name", "place_id"]]
    .rename(columns={"place_id":"canonical_place_id"})
)

display(canonical_place.head())
```

# **Membuat Mapping Place ID**

```python
mapping = (
    df_clean[
        ["place_name","place_id"]
    ]
    .drop_duplicates()
    .merge(
        canonical_place,
        on="place_name",
        how="left"
    )
)

display(mapping.head(20))
```

# **Mengganti Seluruh place_id**

```python
df_clean = df_clean.merge(
    canonical_place,
    on="place_name",
    how="left"
)

df_clean["place_id"] = df_clean["canonical_place_id"]

df_clean.drop(
    columns="canonical_place_id",
    inplace=True
)

print("="*60)
print("NORMALISASI PLACE ID SELESAI")
print("="*60)
```

# **Menampilkan Jumlah place_id yang Dinormalisasi**

```python
jumlah_normalisasi = (
    mapping["place_id"] != mapping["canonical_place_id"]
).sum()

print("="*60)
print("HASIL NORMALISASI PLACE ID")
print("="*60)

print(f"Jumlah place_id yang dinormalisasi : {jumlah_normalisasi:,}")
```

# **Menghapus Duplikasi User-Item**

```python
jumlah_sebelum = len(df_clean)

df_clean = (
    df_clean
    .sort_values(
        by=["review_date","rating"],
        ascending=[False,False]
    )
    .drop_duplicates(
        subset=["user_id","place_id"],
        keep="first"
    )
    .reset_index(drop=True)
)

jumlah_sesudah = len(df_clean)

print("="*60)
print("PENGHAPUSAN DUPLIKASI USER-ITEM")
print("="*60)

print(f"Jumlah data sebelum : {jumlah_sebelum:,}")
print(f"Jumlah data sesudah : {jumlah_sesudah:,}")
print(f"Jumlah data dihapus : {jumlah_sebelum-jumlah_sesudah:,}")
```

# **Menghapus Duplikasi Baris**

```python
jumlah_sebelum = len(df_clean)

df_clean = (
    df_clean
    .drop_duplicates()
    .reset_index(drop=True)
)

jumlah_sesudah = len(df_clean)

print("="*60)
print("PENGHAPUSAN DUPLIKASI BARIS")
print("="*60)

print(f"Jumlah data sebelum : {jumlah_sebelum:,}")
print(f"Jumlah data sesudah : {jumlah_sesudah:,}")
print(f"Jumlah data dihapus : {jumlah_sebelum-jumlah_sesudah:,}")
```

# **Validasi Hasil Praproses**

```python
print("="*70)
print("VALIDASI HASIL PRAPROSES")
print("="*70)

print(f"Jumlah Missing Value : {df_clean.isnull().sum().sum():,}")

print(f"Duplikasi Baris      : {df_clean.duplicated().sum():,}")

cek_user_item = (
    df_clean
    .groupby(["user_id","place_id"])
    .size()
    .gt(1)
    .sum()
)

print(f"User-Item Ganda      : {cek_user_item:,}")

cek_place = (
    df_clean
    .groupby("place_name")["place_id"]
    .nunique()
    .gt(1)
    .sum()
)

print(f"Place Name >1 ID     : {cek_place:,}")

print(f"Jumlah Data Akhir    : {len(df_clean):,}")
```
